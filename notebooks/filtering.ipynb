```python
import json

raw = json.load(open("/Users/dolor.src/Downloads/products_final.json"))
data = raw["products"]
```

```python
from maxidom_hack.db.annotater import DBAnnotater
from maxidom_hack.utils.clients import JudgeClient

annotater = DBAnnotater(JudgeClient())
```

```python
ann_data = await annotater.annotate_async(data)
```

```python
len(ann_data)
```

```python
import json

with open('/Users/dolor.src/code/maxidom-hack/data/ann_data_v1.json', 'w') as f:
    json.dump(ann_data, f, ensure_ascii=False, indent=2)
```

```python
import numpy as np
cats = []
for item in ann_data:
    cats.append(list(item["categories"].values())[0])

ul, uc = np.unique(cats, return_counts=True)

for l, c in zip(ul, uc):
    print(f"{l}: {c}")
```
